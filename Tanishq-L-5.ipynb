Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic 
operations like: 
a) Addition of 2 in all the element  
b) Multiply 3 with all the elements 
c) Divide every element by 2 

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

print("After adding 2:", arr + 2)

print("After multiplying by 3:", arr * 3)

print("After dividing by 2:", arr / 2)
```

Q.2 Questions on Basic NumPy Array: 
a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5]) 
b) Find the most frequent value and their indice(s) in the following arrays: 
i. x = np.array([1,2,3,4,5,1,2,1,1,1]) 
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("(a) Original array:", arr)
print("(a) Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = np.max(counts)
most_frequent = values[counts == max_count]

print("\n(b)(i) Most frequent value:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    print("Indices of", value, ":", np.where(x == value)[0])

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = np.max(counts)
most_frequent = values[counts == max_count]

print("\n(b)(ii) Most frequent values:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    print("Indices of", value, ":", np.where(y == value)[0])
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly 
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, 
shape, total elements, the data type of each element and total number of bytes consumed 
by the array. Find the transpose of this array using reshape() attribute. Can we do the same 
with T attribute? 

```python
import numpy as np

Tanishq = np.linspace(10, 100, 25)

print("Array:")
print(Tanishq)

print("\nDimensions:", Tanishq.ndim)
print("Shape:", Tanishq.shape)
print("Total elements:", Tanishq.size)
print("Data type:", Tanishq.dtype)
print("Total bytes:", Tanishq.nbytes)

transpose = Tanishq.reshape(1, 25)

print("\nTranspose using reshape():")
print(transpose.T)

print("\nUsing T attribute:")
print(transpose.T)

print("\nCan transpose be done using T attribute?")
print("Yes, but a 1-D array's .T does not change its shape.")
print("Therefore, reshape(1, 25).T is used to get a 25 x 1 transpose.")
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> 
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, 
median, max, min, unique elements. Reshape the array to four rows and three columns and 
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three 
columns and name it as resized_ ucs420_<your_name>>.

```python
import numpy as np

ucs420_Tanishq = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Tanishq)

print("\nMean:", np.mean(ucs420_Tanishq))
print("Median:", np.median(ucs420_Tanishq))
print("Maximum:", np.max(ucs420_Tanishq))
print("Minimum:", np.min(ucs420_Tanishq))
print("Unique elements:", np.unique(ucs420_Tanishq))

reshaped_ucs420_Tanishq = ucs420_Tanishq.reshape(4, 3)

print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Tanishq)

resized_ucs420_Tanishq = np.resize(ucs420_Tanishq, (2, 3))

print("\nResized array (2 x 3):")
print(resized_ucs420_Tanishq)
```
