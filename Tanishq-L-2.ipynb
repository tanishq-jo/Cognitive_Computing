```python

roll_no = "1024170156"
L = [int(digit) * 10 for digit in roll_no]

print("Initial L:", L)


L.append(80)
print("After append(80):", L)

L.insert(2, 30)
print("After insert(2, 30):", L)



L.remove(0)
print("After remove(0):", L)


L.pop()
print("After pop():", L)




L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)



print("First three elements:", L[:3])
print("Last three elements:", L[-3:])



average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

```python

L = [10, 0, 20, 40, 10, 70, 0, 10, 50, 60]

scores = tuple(L[:8])

print("Scores:", scores)

highest = max(scores)
highest_index = scores.index(highest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Lowest score:", lowest)
print("Lowest score appears:", lowest_count, "times")


reversed_scores = list(reversed(scores))

print("Reversed tuple as list:", reversed_scores)


score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score not present in the tuple.")


# Attempt to change an element
#scores[0] = 100

# This raises TypeError because tuples are immutable,
# whereas list elements can be changed.

first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
import random

random.seed(1024170156)

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)



odd_numbers = [num for num in numbers if num % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Number of odd numbers:", len(odd_numbers))



even_numbers = [num for num in numbers if num % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Number of even numbers:", len(even_numbers))



def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


prime_numbers = [num for num in numbers if is_prime(num)]

print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))


most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python

digits = [1, 0, 2, 4, 1, 7, 0, 1, 5, 6]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)


union = A.union(B)

print("\nUnion of A and B:", union)


intersection = A.intersection(B)

print("Intersection of A and B:", intersection)


A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)


symmetric_diff = A.symmetric_difference(B)

print("Symmetric difference:", symmetric_diff)

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))


X = int(input("\nEnter a value X: "))

A.discard(X)

print("Set A after discard:", A)

# discard() is safer than remove() because it does not raise an error
# if the value is not present in the set.
```

```python

my_dict = {
    "name": "Tanishq Sharma",
    "roll_no": "1024170156",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)


my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)


my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)


my_dict["age"] += 1

print("\nAfter increasing age:")
print(my_dict)


dict_pop = my_dict.copy()
dict_del = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_branch)

del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value,
# while del removes the key but does not return its value.


print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)


print("\nChecking for email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present.")


friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170200",
    "branch": "ECE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary wins.


string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nOnly key-value pairs having string values:")
print(string_values)
```
